```python
order_raw.groupby('category')['amount'].sum()
```

```python
# Total amount spent per category
category_summary = order_raw.groupby('category')['amount'].sum()
print(category_summary)
```

```python
# Remove duplicate rows from order_raw
orders = order_raw.drop_duplicates()
```

```python
# Check which columns have NaN values
print(order_raw.isna().sum())
```

```python
# Option A: Fill missing numerical ratings with the average rating
order_raw['rating'] = order_raw['rating'].fillna(order_raw['rating'].mean())
```

```python
# Option B: Fill missing priority with 'Not Specified'
order_raw['priority'] = order_raw['priority'].fillna('Not Specified')
```

```python

# Merge orders with customer profile data
merged_df = pd.merge(order_raw, Customer, on='customer_id', how='left')
```
